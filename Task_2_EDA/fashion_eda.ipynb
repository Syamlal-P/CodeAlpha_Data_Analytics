## Step 7: Hypothesis Testing

### Hypothesis 1:
Higher-priced fashion products are less common than mid-range products.

### Hypothesis 2:
The dataset does not contain significant price outliers.

### Hypothesis 3:
Product prices follow a normal or skewed distribution.


```python
plt.figure()
sns.histplot(df["price"], bins=5, kde=True)
plt.title("Price Frequency Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

Mid-range prices appear more frequently than extreme low or high prices, supporting Hypothesis 1.


```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["price"] < lower_bound) | (df["price"] > upper_bound)]
outliers
```

The dataset contains very few or no extreme outliers, supporting Hypothesis 2.


```python
plt.figure()
sns.kdeplot(df["price"], fill=True)
plt.title("Kernel Density Estimation of Price Distribution")
plt.xlabel("Price")
plt.show()
```

The price distribution is slightly skewed, indicating non-normal distribution of fashion product prices.
